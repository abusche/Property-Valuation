```python
from selenium import webdriver
from selenium.webdriver.firefox.service import Service
from selenium.webdriver.firefox.options import Options
from webdriver_manager.firefox import GeckoDriverManager
from selenium.webdriver.common.by import By
from selenium.webdriver.common.action_chains import ActionChains
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from tqdm import tqdm
import time
from bs4 import BeautifulSoup
```

```python
def get_soup(url):
    options = Options()
    options.add_argument("--headless")
    driver = webdriver.Firefox(service=Service(GeckoDriverManager().install()), options=options)
    driver.get(url)

    try:
        time.sleep(2)
        wait = WebDriverWait(driver, 5)
        
        buttons = driver.find_elements(By.CSS_SELECTOR, "button[data-qa-id='criteria_more']")
        if not buttons:
            print("Bouton non trouvé, récupération du HTML actuel.")
        else:
            button = buttons[0]
            #print("Bouton trouvé...")

            driver.execute_script("arguments[0].scrollIntoView({behavior: 'smooth', block: 'center'});", button)
            time.sleep(1)

            #print("Affichage du bouton...")
            driver.execute_script("arguments[0].style.display = 'block'; arguments[0].disabled = false;", button)
            time.sleep(1)

            try:
                button.click()
                #print("Clic réussi")
            except Exception:
                print("Clic échoué, tentative avec ActionChains...")

            time.sleep(2)
    except Exception as e:
        print(f"Erreur lors du clic sur le bouton: {e}")
    finally:
        html = driver.page_source
        driver.quit()
    
    soup = BeautifulSoup(html, "html.parser")
    return soup


def get_links(url):
    end = False
    i = 1
    while end == True:
        if i == 1:
            path = url
            i += 1
        else:
            i += 1
            path = url + f"&page={i}"
        soup = get_soup(path)
        codes = soup.find_all('a', class_='absolute inset-none')
        links = []
        for code in codes:
            links.append(f"https://www.leboncoin.fr" + code.get('href'))
    
    return links
```

```python
def get_all_links(url, max_pages=5):
    options = Options()
    options.add_argument("--headless")
    driver = webdriver.Firefox(service=Service(GeckoDriverManager().install()), options=options)
    driver.get(url)
    
    all_links = []
    
    try:
        for page in range(max_pages):
            # Debugging : Vérifier les éléments avant l'attente
            elements = driver.find_elements(By.CSS_SELECTOR, "a.absolute.inset-none")
            print(f"Page {page + 1}: {len(elements)} éléments trouvés avant l'attente.")
            
            # Attendre que les annonces soient chargées
            wait = WebDriverWait(driver, 30)  # Réinitialiser le wait pour chaque page
            wait.until(EC.presence_of_all_elements_located((By.CSS_SELECTOR, "a.absolute.inset-none")))
            
            # Récupérer les liens de la page actuelle
            soup = BeautifulSoup(driver.page_source, "html.parser")
            links = soup.find_all('a', class_='absolute inset-none')
            page_links = [link['href'] for link in links if 'href' in link.attrs]
            all_links.extend(page_links)
            
            print(f"Page {page + 1}: {len(page_links)} liens récupérés.")
            
            # Vérifier si le bouton "Page suivante" existe
            try:
                next_button = wait.until(EC.presence_of_element_located((By.CSS_SELECTOR, "a[aria-label='Page suivante']")))
                if next_button:
                    print("Bouton 'Page suivante' trouvé.")
                    
                    # Défilement vers le bouton et clic
                    driver.execute_script("arguments[0].scrollIntoView({behavior: 'smooth', block: 'center'});", next_button)
                    time.sleep(1)
                    driver.execute_script("arguments[0].click();", next_button)
                    
                    # Attendre que l'URL change ou que de nouveaux éléments se chargent
                    wait.until(EC.staleness_of(next_button))  # Attendre que le bouton "Page suivante" devienne obsolète
                    wait.until(EC.presence_of_all_elements_located((By.CSS_SELECTOR, "a.absolute.inset-none")))
                    time.sleep(5)  # Donne un délai supplémentaire pour le chargement
                else:
                    print("Bouton 'Page suivante' introuvable ou fin de pagination atteinte.")
                    break
            except Exception as e:
                print(f"Erreur : {e}. Bouton 'Page suivante' introuvable ou fin de pagination atteinte.")
                break
    finally:
        driver.quit()
    
    return all_links
```

```python
# Exemple d'utilisation
url = "https://www.leboncoin.fr/recherche?category=10&locations=Strasbourg__48.60684920306528_7.693333916606193_10000_5000&price=min-1000&rooms=2-3&square=50-max&real_estate_type=2"
links = get_all_links(url, max_pages=5)

print(f"Nombre total de liens récupérés : {len(links)}")
for link in links:
    print(link)
```

```python
def get_piece(soup):
    piece = soup.find_all("span", class_="text-body-1")[0].text.replace("\xa0·\xa0", "")
    return piece


def get_m2(soup):
    m2 = soup.find_all("span", class_="text-body-1")[1].text.replace("\xa0·\xa0", "")
    return m2

def get_ville_dep(soup):
    code = soup.find_all("a", class_="text-body-1")
    return code[len(code)-1].text.replace("\xa0", "")


def get_description(soup):
    desc = soup.find_all("div", class_="mx-lg grid gap-lg py-xl border-b-sm border-b-neutral/dim-4 lg:mx-md row-start-1")[0].text
    return desc

def get_criteres(soup):
    criteres = []
    codes = soup.find_all("p", class_="truncate text-body-1 font-bold")
    for code in codes:
        criteres.append(code.text)

    return criteres

def get_nrj(soup):
    nrj = []
    codes = soup.find_all("div", class_="flex h-sz-16 items-center justify-center px-sm font-semi-bold text-surface box-content h-sz-24 rounded-md border-md border-solid border-surface text-body-2 drop-shadow")
    for code in codes:
        nrj.append(code.text)
    
    return nrj


def get_type_annonce(soup):
    type_annonce = soup.find_all("span", itemprop="name")[1].text
    return type_annonce


def get_region(soup):
    region = soup.find_all("span", itemprop="name")[2].text
    return region


def get_departement(soup):
    departement = soup.find_all("span", itemprop="name")[3].text
    return departement


def get_loc(soup):
    loc = soup.find_all("span", itemprop="name")[4].text
    return loc


def get_data(soup):
    data = []
    data.append(get_type_annonce(soup))
    data.append(get_ville_dep(soup))
    data.append(get_region(soup))
    data.append(get_departement(soup))
    data.append(get_loc(soup))
    criteres = get_criteres(soup)
    for critere in criteres:
        data.append(critere)
    data.append(get_description(soup))
    return data
```

```python
def get_criteres_name(soup):
    criteres_name = []
    codes = soup.find_all("p", class_="mb-sm truncate text-caption opacity-dim-1")
    for code in codes:
        criteres_name.append(code.text)

    return criteres_name


def get_nrj_name(soup):
    nrj_name = []
    codes = soup.find_all("p", class_="mb-sm flex text-caption opacity-dim-1")
    for code in codes:
        nrj_name.append(code.text.replace("En savoir plus", ""))
    
    return nrj_name


def obtain_list_criteres(links):
    criteres_list = []
    for link in tqdm(links):
        print(link)
        soup = get_soup(link)
        criteres_name = get_criteres_name(soup)
        for critere_name in criteres_name:
            if critere_name not in criteres_list:
                print(cretere_name)
                criteres_list.append(critere_name)
    
    return criteres_list
```

```python
options = Options()
options.add_argument("--headless")
driver = webdriver.Firefox(service=Service(GeckoDriverManager().install()), options=options)
url = "https://www.leboncoin.fr/recherche?category=9&locations=Strasbourg__48.60684920306528_7.693333916606193_10000_5000"

links = get_links(url, driver)
print("Liens obtenus")
list_criteres = obtain_list_criteres(links, driver)
driver.quit()
list_criteres
```
